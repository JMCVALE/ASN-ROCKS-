Ajustando o modelo 1, para C=10

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm

# Criando 60 pontos linearmente separáveis
np.random.seed(0)
X = np.r_[np.random.randn(30, 2) - [1, 1],
np.random.randn(30, 2) + [2, 2]]
Y = [0] * 30 + [1] * 30

# número da figura
fignum = 1

# Ajustando o modelo 1, para C=10
clf = svm.SVC(kernel='linear', C=10)
clf.fit(X, Y)

# Ajustando o hiperplano
w = clf.coef_[0]
a = -w[0] / w[1]
xx = np.linspace(-5, 5)
yy = a * xx - (clf.intercept_[0]) / w[1]

# Trançado o Hiperplano e margens
margin = 1 / np.sqrt(np.sum(clf.coef_ ** 2))
yy_down = yy - np.sqrt(1 + a ** 2) * margin
yy_up = yy + np.sqrt(1 + a ** 2) * margin

# plot das linhas, pontos e vetores de suporte
plt.figure(fignum, figsize=(4, 3))
plt.clf()
plt.plot(xx, yy, 'k-')
plt.plot(xx, yy_down, 'k--')
plt.plot(xx, yy_up, 'k--')

plt.scatter(clf.support_vectors_[:, 0], clf.support_vectors_[:, 1], s=80,
            facecolors='none', zorder=10, edgecolors='k')
plt.scatter(X[:, 0], X[:, 1], c=Y, zorder=10, cmap=plt.cm.Paired,
           edgecolors='k')

plt.axis('tight')
x_min = -4.8
x_max = 4.2
y_min = -6
y_max = 6

XX, YY = np.mgrid[x_min:x_max:200j, y_min:y_max:200j]
Z = clf.predict(np.c_[XX.ravel(), YY.ravel()])

# Criando imagem final
Z = Z.reshape(XX.shape)
plt.figure(fignum, figsize=(4, 3))
plt.pcolormesh(XX, YY, Z, cmap=plt.cm.Paired)

plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)

plt.xticks(())
plt.yticks(())
fignum = fignum + 1

plt.show()
```
